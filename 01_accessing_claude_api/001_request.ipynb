```python
# Load env variables
from dotenv import load_dotenv

load_dotenv(override=True)
```

```python
# Make sure api was loaded
import os

api_key = os.environ.get("ANTHROPIC_API_KEY")
```

```python
# Create an API client

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
def add_user_messages(messages, text):
    user_messages = {"role": "user", "content": text}
    messages.append(user_messages)

def add_assistant_messages(messages, text):
    assistant_messages = {"role": "assistant", "content": text}
    messages.append(assistant_messages)

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens = 1000,
        messages = messages
    )
    return message.content[0].text
```

```python
# Start with an empty message list
messages = []

# Add the initial user question
add_user_messages(messages, "Define quatum computing in one sentence")

# Get Claude's response
answer = chat(messages)

# Add Claude's response to the conversation history
add_assistant_messages(messages, answer)

# Add a follow-up question
add_user_messages(messages, "Write another sentence")

# Get the follow-up response with full context
answer = chat(messages)
```
