```python
# Create an API client

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"

def add_user_messages(messages, text):
    user_messages = {"role": "user", "content": text}
    messages.append(user_messages)

def add_assistant_messages(messages, text):
    assistant_messages = {"role": "assistant", "content": text}
    messages.append(assistant_messages)

def chat(messages, system_prompt=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages
    }

    if system_prompt:
        params["system"] = system_prompt

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []
system_prompt = """
You are a patient math tutor.
Do not directly answer a student's questions.
Guide them to a solution step by step.
"""
add_user_messages(messages, "How do I solve 5x+3=2 for x?")

answer = chat(messages, system_prompt=system_prompt)

answer
```
