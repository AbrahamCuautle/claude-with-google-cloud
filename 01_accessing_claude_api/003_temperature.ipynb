```python
# Create an API client

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"

def add_user_messages(messages, text):
    user_messages = {"role": "user", "content": text}
    messages.append(user_messages)

def add_assistant_messages(messages, text):
    assistant_messages = {"role": "assistant", "content": text}
    messages.append(assistant_messages)

def chat(messages, system_prompt=None, temperature=1.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "extra_body": {
            "temperature": temperature,
            "thinking": {
                "type": "disabled"
            }
        }
    }

    if system_prompt:
        params["system"] = system_prompt

    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []
add_user_messages(
    messages, 
    "Generate a one sentence movie idea"
)
answer = chat(messages, temperature=1.0)
answer
```
