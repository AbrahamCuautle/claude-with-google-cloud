````python
# Create an API client

from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"

def add_user_messages(messages, text):
    user_messages = {"role": "user", "content": text}
    messages.append(user_messages)

def add_assistant_messages(messages, text):
    assistant_messages = {"role": "assistant", "content": text}
    messages.append(assistant_messages)

def chat(messages, system_prompt=None, temperature=1.0, stop_sequences=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "extra_body": {
            "temperature": temperature,
            "thinking": {
                "type": "disabled"
            }
        },
        "stop_sequences": stop_sequences
    }

    if system_prompt:
        params["system"] = system_prompt

    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    message = client.messages.create(**params)
    return message.content[0].text
````

````python
# 1. Creamos la lista de mensajes terminando estrictamente con el usuario
messages = []
add_user_messages(messages, "Generate a very short event bridge rule as json. Start your response directly with the opening JSON object.")

# 2. Usamos el System Prompt para forzar las reglas de formato y el stop sequence
system_prompt = "You are a JSON generator. Output ONLY valid raw JSON code. Do not include markdown blocks like ```json."

# 3. Llamamos a la función (el modelo se detendrá cuando cierre el JSON con '}')
answer = chat(
    messages=messages, 
    system_prompt=system_prompt,
    stop_sequences=["}"]
)

# Añadimos manualmente el cierre de la llave que el stop_sequence cortó
full_json = answer + "}"
print(full_json)
````
